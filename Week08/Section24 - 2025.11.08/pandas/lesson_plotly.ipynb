# What is Matplotlib?

Matplotlib is a comprehensive visualization library used for creating 2D and 3D plots in Python. Developed by John Hunter in 2003, it remains the most widely used visualization tool in the Python ecosystem today.


## Key Features

* MATLAB-like interface

* High-quality plot generation

* Support for numerous plot types

* Extensive customization options

* LaTeX support

* Interactive plotting capabilities

## Why Use It?

* Advantages:
    1. Easy to Learn: MATLAB-like syntax

    2. Flexibility: Ability to customize any type of plot

    3. Wide Adoption: From scientific research to industrial applications

    4. Integration: Compatible with libraries like NumPy and Pandas

    5. Cross-Platform: Works on Windows, Linux, and macOS

* Use Cases:

    1. Data analysis and visualization

    2. Scientific research

    3. Visualizing machine learning results

    4. Financial data analysis

    5. Academic publications

# What is Plotly?

Plotly is a powerful, interactive graphing library for Python. Unlike Matplotlib, which produces static images by default, Plotly creates interactive plots that allow users to zoom, pan, and hover over data points.


## Key Features

* Interactive by default

* Web-based output (HTML/JSON)

* High-level API (Plotly Express) and Low-level API (Graph Objects)

* Extensive support for 3D and geographic plots

* Easy integration with Dash for building analytical dashboards

* Seamless compatibility with Polars and Pandas

## Why Use It?

* Advantages:
    1. Interactivity: Explore data dynamically without re-rendering plots

    2. Modern Aesthetics: Polished, professional look out of the box

    3. Versatility: From simple charts to complex scientific visualizations

    4. Web-Ready: Perfectly suited for sharing via browser or embedding in apps

    5. Declarative Syntax: Express complex plots with very little code

* Use Cases:

    1. Interactive data exploration

    2. Business intelligence dashboards

    3. Scientific data visualization

    4. Financial time-series analysis
    5. Geographic mapping and spatial data analysis

## Line Plot

# Figure and Layout Concepts (Declarative Approach)

Plotly's core philosophy relies on the Figure object. A Figure contains data (traces) and layout settings. This approach provides full control over the plot.

| Concept | Definition | Analogy |
|---------|--------|----------|
| Figure | The overall object containing data and layout. | A complete framed picture. |
| Traces | The actual data being plotted (e.g., a line or scatter). | The drawing on the paper. |
| Layout | The visual settings like titles, axes, and legends. | The frame and labels around the picture. |

````python
# 1. Create Figure
fig = go.Figure()

# 2. Prepare Data
x_data = [1, 2, 3, 4]
y_data = [10, 20, 35, 25]

# 3. Draw the Plot
# Use add_trace to add a line with markers
fig.add_trace(
    go.Scatter(
        x=x_data,
        y=y_data,
        mode="lines+markers",
        line=dict(color="red", width=2, dash="dot"),
        marker=dict(size=8),
    )
)

fig.update_layout(title="Simple Line Plot", xaxis_title="X Axis", yaxis_title="Y Axis")

# 4. Display the Plot
fig.show()
````

# Creating Your First Line Plot (px.line())

Let's create a basic line plot using px.line() and configure basic settings. For simplicity, we can demonstrate the Plotly Express approach here (single-line plotting).

````python
# Simple line plot - Plotly Express method

# Create dataset (e.g., sine wave)
x = np.linspace(0, 10, 100)  # 100 evenly spaced points from 0 to 10
y = np.sin(x)

# Single line plotting
fig = px.line(x=x, y=y)
fig.show()
````

````python
# Line Customization
fig = go.Figure()

# Customize the line
fig.add_trace(
    go.Scatter(
        x=x,
        y=y,
        mode="lines",
        line=dict(color="red", width=2, dash="dash"),
        name="Sine Wave",
    )
)

# Add a second line (Cosine)
fig.add_trace(
    go.Scatter(
        x=x,
        y=np.cos(x),
        mode="lines",
        line=dict(color="blue", width=1),
        name="Cosine Wave",
    )
)

fig.show()
````

# Customizing and Labeling the Plot

Adding titles, axis labels, and legends to improve plot readability.

````python
# Data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Draw the plot
fig = px.line(
    x=x,
    y=y,
    title="Sine Function Over Time",
    labels={"x": "Time (t)", "y": "Amplitude (A)"},
)

# 1. Customize title and axis labels
fig.update_layout(
    title=dict(text="Sine Function Over Time", font=dict(size=16, color="darkblue")),
    xaxis=dict(title="Time (t)", title_font=dict(size=12)),
    yaxis=dict(title="Amplitude (A)", title_font=dict(size=12)),
)

# 2. Add Legend (Managed automatically by Plotly for named traces)
fig.update_layout(legend=dict(x=0, y=1, xanchor="left", yanchor="top"))

# 3. Set Axis Limits
fig.update_xaxes(range=[0, 10])
fig.update_yaxes(range=[-1.5, 1.5])

# 4. Add Grid Lines
fig.update_xaxes(showgrid=True, gridwidth=1, gridcolor="LightGray")
fig.update_yaxes(showgrid=True, gridwidth=1, gridcolor="LightGray")

fig.show()
````

# Subplots

Multiple Plot Layouts (Subplots)


Description: Often we need to draw multiple plots in a single window to examine different aspects of data simultaneously. In Plotly, we use the `make_subplots` utility.

Grid Layout with `make_subplots()`
Example: 2x2 Grid Layout

````python
from plotly.subplots import make_subplots

# Create 4 Axes in 2 rows and 2 columns
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        "Linear (y=x)",
        "Quadratic (y=x^2)",
        "Square Root (y=sqrt(x))",
        "Cubic (y=x^3)",
    ),
)

# Data
x = np.linspace(0, 5, 50)
y1 = x
y2 = x**2
y3 = np.sqrt(x)
y4 = x**3

# Top Left (1, 1)
fig.add_trace(go.Scatter(x=x, y=y1, mode="lines", line=dict(color="red")), row=1, col=1)

# Top Right (1, 2)
fig.add_trace(
    go.Scatter(x=x, y=y2, mode="lines", line=dict(color="green")), row=1, col=2
)

# Bottom Left (2, 1)
fig.add_trace(
    go.Scatter(x=x, y=y3, mode="lines", line=dict(color="blue")), row=2, col=1
)

# Bottom Right (2, 2)
fig.add_trace(
    go.Scatter(x=x, y=y4, mode="lines", line=dict(color="purple")), row=2, col=2
)

fig.update_layout(title_text="Various Mathematical Functions", height=800, width=1000)
fig.show()
````

Asymmetric Layouts with make_subplots()
Sometimes asymmetric layouts are needed. In such cases, we define the grid specs.

````python
from plotly.subplots import make_subplots

# Create a layout where the bottom plot spans both columns
fig = make_subplots(rows=2, cols=2, specs=[[{}, {}], [{"colspan": 2}, None]])

fig.add_trace(go.Scatter(x=x, y=y1, name="Plot 1"), row=1, col=1)
fig.add_trace(go.Scatter(x=x, y=y2, name="Plot 2"), row=1, col=2)
fig.add_trace(go.Scatter(x=x, y=y3, name="Plot 3"), row=2, col=1)

fig.update_layout(height=600, width=900, title_text="Asymmetric Layout")
fig.show()
````

````python
# 2x2 subplot layout
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        "Sine Function",
        "Cosine Function",
        "Tangent Function",
        "Exponential Function",
    ),
)

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(x / 10)

fig.add_trace(go.Scatter(x=x, y=y1, line=dict(color="red")), row=1, col=1)
fig.add_trace(go.Scatter(x=x, y=y2, line=dict(color="blue")), row=1, col=2)
fig.add_trace(go.Scatter(x=x, y=y3, line=dict(color="green")), row=2, col=1)
fig.add_trace(go.Scatter(x=x, y=y4, line=dict(color="magenta")), row=2, col=2)

fig.update_yaxes(range=[-5, 5], row=2, col=1)
fig.update_layout(height=800, width=1000, showlegend=False)
fig.show()
````

````python
# Detailed customization example
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=x,
        y=y1,
        name="sin(x)",
        line=dict(color="red", width=1),
        marker=dict(symbol="circle", size=4),
        mode="lines+markers",
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=y2,
        name="cos(x)",
        line=dict(color="blue", width=1, dash="dash"),
        marker=dict(symbol="square", size=4),
        mode="lines+markers",
    )
)

fig.update_layout(
    title=dict(text="Trigonometric Functions", font=dict(size=16, color="black")),
    xaxis=dict(title="X Values", title_font=dict(size=14), range=[0, 2 * np.pi]),
    yaxis=dict(title="Y Values", title_font=dict(size=14), range=[-1.5, 1.5]),
    plot_bgcolor="white",
)

# Add grid and spines (axes lines)
fig.update_xaxes(
    showline=True,
    linewidth=2,
    linecolor="black",
    mirror=False,
    showgrid=True,
    gridcolor="LightGray",
)
fig.update_yaxes(
    showline=True,
    linewidth=2,
    linecolor="black",
    mirror=False,
    showgrid=True,
    gridcolor="LightGray",
)

# Reference lines
fig.add_hline(y=0, line_color="black", opacity=0.3)
fig.add_vline(x=np.pi, line_color="gray", line_dash="dot", opacity=0.5)

fig.show()
````

# Animations

````python
# Plotly animations are typically created using px.line with animation_frame
import pandas as pd  # Using pandas for animation frame structure as it is more standard for Plotly Express animations

x = np.linspace(0, 4 * np.pi, 100)
frames = 100
data = []
for f in range(frames):
    y = np.sin(x + f / 10)
    for xi, yi in zip(x, y):
        data.append([xi, yi, f])

df_anim = pd.DataFrame(data, columns=["x", "y", "frame"])

fig = px.line(
    df_anim,
    x="x",
    y="y",
    animation_frame="frame",
    range_y=[-1.5, 1.5],
    title="Sine Wave Animation",
)
fig.show()
````

# 5.2 Heatmap

Used to visualize correlation between multiple numerical variables in a single matrix.

````python
# Select only numerical columns and compute correlation matrix
numeric_df = df_tips.select(pl.col(pl.Float64, pl.Int64))
cols = numeric_df.columns
corr_matrix = []
for col1 in cols:
    row = [numeric_df[col1].corr(numeric_df[col2]) for col2 in cols]
    corr_matrix.append(row)

fig = px.imshow(
    corr_matrix,
    x=cols,
    y=cols,
    text=corr_matrix,
    color_continuous_scale="Viridis",
    title="Correlation Matrix of Numerical Variables",
)
fig.show()
````

````python
# Correlation heatmap
iris_num = px.data.iris().select(pl.col(pl.Float64, pl.Int64))
cols = iris_num.columns
corr_matrix = [[iris_num[c1].corr(iris_num[c2]) for c2 in cols] for c1 in cols]

fig = px.imshow(
    corr_matrix,
    x=cols,
    y=cols,
    text=corr_matrix,
    color_continuous_scale="RdBu",
    title="Iris Dataset - Correlation Matrix",
)
fig.show()
````

# 7.3 Setting Plot Style

Plotly offers ready-made templates for visual elements like background and grid lines.

## 🎨 Plotly Style Settings

| Template | Description |
|----------|-----------|
| `plotly_white` | White background with clean lines. |
| `plotly_dark` | Dark background for high contrast. |
| `ggplot2` | Mimics the style of R's ggplot2. |
| `seaborn` | Mimics the style of Python's Seaborn. |
| `simple_white` | Minimalist white background. |

````python
# Change plot style (e.g., white grid)
px.set_template("plotly_white")

# Draw example Box Plot
fig1 = px.box(data=df_tips, x="day", y="total_bill", title="Box Plot with White Style")
fig1.show()

# Change style and draw another plot
px.set_template("ggplot2")
fig2 = px.scatter(
    data=df_tips, x="total_bill", y="tip", title="Scatter Plot with ggplot2 Style"
)
fig2.show()

# Return to standard style
px.set_template("plotly")
````
