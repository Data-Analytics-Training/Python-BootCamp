## Line Plot

````python
# Simple line plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.figure(figsize=(8, 6))
plt.plot(x, y, marker="o", linestyle=":", color="blue", linewidth=2, markersize=8)
plt.title("Simple Line Plot", fontsize=14)
plt.xlabel("X Axis", fontsize=12)
plt.ylabel("Y Axis", fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
````

# Figure and Axes Concepts (OOP Approach)

Matplotlib's core philosophy relies on two main components: Figure and Axes. This Object-Oriented Programming (OOP) approach provides full control over the plot.

| Concept | Definition | Analogy |
|---------|--------|----------|
| Figure | The overall window or canvas where plots are drawn. Can contain multiple Axes. | A picture frame or a blank sheet of paper. |
| Axes | The actual area where the plot is drawn. Where X and Y axes are defined. Each Figure can contain multiple Axes. | The actual drawing area inside the frame. |

````python
# 1. Create Figure and Axes
# fig: Figure object (canvas)
# ax: Axes object (plotting area)
fig, ax = plt.subplots(figsize=(8, 4))

# 2. Prepare Data
x_data = [1, 2, 3, 4]
y_data = [10, 20, 35, 25]

# 3. Draw the Plot
# Use ax.plot() method to draw on the Axes object
ax.plot(
    x_data, y_data, marker="o", linestyle=":", color="red", linewidth=2, markersize=8
)

ax.set_title("Simple Line Plot", fontsize=14)
ax.set_xlabel("X Axis", fontsize=12)
ax.set_ylabel("Y Axis", fontsize=12)
ax.grid(True, alpha=0.3, linestyle="--", color="yellow")

# 4. Display the Plot (Opens window or renders output)
plt.show()
````

# Creating Your First Line Plot (`plt.plot()`)

Let's create a basic line plot using plt.plot() and configure basic settings. For simplicity, we can also demonstrate the Procedural approach here (single-line plotting).

````python
# Simple line plot - procedural method

# Create dataset (e.g., sine wave)
x = np.linspace(0, 10, 100)  # 100 evenly spaced points from 0 to 10
y = np.sin(x)

# Single line plotting (Matlab/plt.plot() style)
# Matplotlib automatically creates a Figure and an Axes in the background.
plt.plot(x, y)
plt.show()
````

````python
# Line Customization
# Back to OOP Approach
fig, ax = plt.subplots(figsize=(8, 4))  # Set figure size with figsize

# Customize the line
# 'r--' -> Red color, dashed line
# linewidth -> Line thickness
# label -> Label for legend
ax.plot(x, y, "r--", linewidth=2, label="Sine Wave")

# Add a second line (Cosine)
ax.plot(
    x, np.cos(x), "b-", linewidth=1, label="Cosine Wave"
)  # 'b-' -> Blue color, solid line

plt.show()
````

# Customizing and Labeling the Plot

Adding titles, axis labels, and legends to improve plot readability.

````python
# Data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create Figure and Axes
fig, ax = plt.subplots(figsize=(10, 5))  # figsize 10x5 inches

# Draw the plot
ax.plot(x, y, color="purple", linestyle="-", label="Sine Plot")

# 1. Add Title
# Adjust font size with fontsize
ax.set_title("Sine Function Over Time", fontsize=16, color="darkblue")

# 2. Add Axis Labels
ax.set_xlabel("Time (t)", fontsize=12)
ax.set_ylabel("Amplitude (A)", fontsize=12)

# 3. Add Legend (Displays labels defined via label parameter)
ax.legend(loc="lower left")  # loc: legend position

# 4. Set Axis Limits
# Defines min/max values displayed on the plot
ax.set_xlim(0, 10)
ax.set_ylim(-1.5, 1.5)

# 5. Add Grid Lines
ax.grid(True, linestyle=":", alpha=0.6)

plt.show()
````

````python
# To visualize the relationship or correlation between two continuous variables.

# Generate random data
np.random.seed(42)  # For reproducibility
height = np.random.normal(170, 10, 50)  # Mean 170, Std Dev 10
weight = height * 0.6 + np.random.normal(5, 5, 50)  # Weight correlated with height

fig, ax = plt.subplots(figsize=(8, 5))

# Draw scatter plot
# s: marker size, c: marker color, alpha: transparency
ax.scatter(height, weight, s=50, c="red", alpha=0.7)

ax.set_title("Height vs Weight Relationship (Scatter Plot)")
ax.set_xlabel("Height (cm)")
ax.set_ylabel("Weight (kg)")
ax.grid(True, linestyle="--")

plt.show()
````

# Bar Plot

````python
# Used to compare categorical data. Categories are on the X-axis,
# and corresponding values (count, average, etc.) are on the Y-axis.

# Command: ax.bar(categories, values)

# Categorical data
products = ["A", "B", "C", "D", "E"]
sales = [250, 400, 150, 350, 500]

fig, ax = plt.subplots(figsize=(8, 5))

# Draw bar plot
ax.bar(products, sales, color=["skyblue", "salmon", "lightgreen", "gold", "grey"])

ax.set_title("Monthly Product Sales (Bar Chart)")
ax.set_xlabel("Product Category")
ax.set_ylabel("Total Units Sold")

plt.show()
````

````python
# Vertical bar chart
categories = ["A", "B", "C", "D", "E"]
values = [23, 45, 56, 78, 33]

plt.figure(figsize=(10, 6))
bars = plt.bar(categories, values, color=["red", "blue", "green", "orange", "purple"])
plt.title("Vertical Bar Chart", fontsize=14)
plt.xlabel("Categories")
plt.ylabel("Values")

# Add value labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2.0,
        height,
        f"{height}",
        ha="center",
        va="bottom",
    )

plt.show()
````

````python
# Horizontal bar chart
# Data
categories = ["A", "B", "C", "D", "E"]
values = [23, 45, 56, 78, 33]

# Figure size
plt.figure(figsize=(10, 6))

# Horizontal bar chart
bars = plt.barh(categories, values, color=["red", "blue", "green", "orange", "purple"])

# Title and labels
plt.title("Horizontal Bar Chart", fontsize=14)
plt.xlabel("Values")
plt.ylabel("Categories")

# Add value labels at the end of bars
for bar in bars:
    width = bar.get_width()  # bar length (value)
    plt.text(
        width + 1,  # place text just right of the bar
        bar.get_y() + bar.get_height() / 2,  # center vertically
        f"{width}",
        va="center",
    )

plt.show()
````

# Subplots

Multiple Plot Layouts (Subplots)


Description: Often we need to draw multiple plots in a single window to examine different aspects of data simultaneously. In Matplotlib, this is called a subplot.

Grid Layout with `plt.subplots()`
`plt.subplots(nrows, ncols)` is the most common method. It automatically returns a Figure and specified number of Axes objects.

Example: 2x2 Grid Layout

````python
# Create 4 Axes (ax) in 2 rows and 2 columns on a single Figure (fig)
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))
# ax is now a 2D NumPy array, accessed via indexing: ax[row, col]

# Data
x = np.linspace(0, 5, 50)
y1 = x
y2 = x ** 2
y3 = np.sqrt(x)
y4 = x ** 3

# Top Left (0, 0)
ax[0, 0].plot(x, y1, "r")
ax[0, 0].set_title("Linear (y=x)")

# Top Right (0, 1)
ax[0, 1].plot(x, y2, "g")
ax[0, 1].set_title("Quadratic (y=x^2)")

# Bottom Left (1, 0)
ax[1, 0].plot(x, y3, "b")
ax[1, 0].set_title("Square Root (y=sqrt(x))")

# Bottom Right (1, 1)
ax[1, 1].plot(x, y4, "purple")
ax[1, 1].set_title("Cubic (y=x^3)")

# Main title for all plots
fig.suptitle("Various Mathematical Functions", fontsize=16)

# Adjust spacing to prevent overlapping
fig.tight_layout(pad=3.0)

plt.show()
````

Asymmetric Layouts with add_subplot()
Sometimes asymmetric layouts are needed. In such cases, each Axes can be added individually.

````python
fig = plt.figure(figsize=(9, 6))

# (num_rows, num_cols, position)
# Plot 1: Occupies full left side (row 1 of 2 rows, 1 col)
ax1 = fig.add_subplot(2, 2, 1)  # Top Left
ax1.plot(x, y1, label="Plot 1")

# Plot 2: Top Right
ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(x, y2, label="Plot 2")

# Plot 3: Occupies full bottom side (row 2 of 2 rows, 1 col)
ax3 = fig.add_subplot(2, 1, 2)
ax3.plot(x, y3, label="Plot 3")

fig.tight_layout()
plt.show()
````

````python
# 2x2 subplot layout
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Generate data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(x / 10)

# Plot 1
axes[0, 0].plot(x, y1, "r-")
axes[0, 0].set_title("Sine Function")
axes[0, 0].grid(True)

# Plot 2
axes[0, 1].plot(x, y2, "b-")
axes[0, 1].set_title("Cosine Function")
axes[0, 1].grid(True)

# Plot 3
axes[1, 0].plot(x, y3, "g-")
axes[1, 0].set_title("Tangent Function")
axes[1, 0].grid(True)
axes[1, 0].set_ylim(-5, 5)

# Plot 4
axes[1, 1].plot(x, y4, "m-")
axes[1, 1].set_title("Exponential Function")
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
````

# 3D Plots

````python
# 3D surface plot
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection="3d")

# Generate data
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X ** 2 + Y ** 2))

# Draw 3D surface
surface = ax.plot_surface(X, Y, Z, cmap="viridis", alpha=0.9)

# Axis labels
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_title("3D Surface Plot")

# Colorbar
fig.colorbar(surface, ax=ax, shrink=0.5, aspect=5)

plt.show()
````

# Animations

````python
from matplotlib.animation import FuncAnimation

# Create figure and axes
fig, ax = plt.subplots(figsize=(8, 6))

# X data (100 points between 0 and 4π)
x = np.linspace(0, 4 * np.pi, 100)

# Initial sine wave
(line,) = ax.plot(x, np.sin(x))

# Fix Y axis limits (otherwise it flickers)
ax.set_ylim(-1.5, 1.5)


# Function called at each frame
def animate(frame):
    line.set_ydata(np.sin(x + frame / 10))  # shift phase
    return (line,)


# Create animation
ani = FuncAnimation(fig, animate, frames=100, interval=50, blit=True)

# Display
plt.show()

ani.save("sine_animation.gif", writer="pillow", fps=20)
````

# 4.4 Histogram and KDE (Kernel Density Estimation)

Seaborn's histplot (similar to Matplotlib's hist) and kdeplot (Kernel Density Estimation) show the shape of the distribution.

Statistically rich plots summarizing distributions of categorical and continuous variables. These plots show quartiles, median, and outliers of a variable.

````python
# Examine distribution of total_bill variable
fig, ax = plt.subplots(figsize=(8, 4))

# Histogram (Seaborn style, more aesthetic than Matplotlib)
sns.histplot(
    data=df_tips,  # Data source (Pandas DataFrame)
    x="total_bill",  # Column for X axis
    bins=15,
    kde=True,  # Add density estimation line
    color="darkorange",
    ax=ax,
)

ax.set_title("Total Bill Distribution (Histplot + KDE)")
plt.show()

# Density Plot Only (KDE)
fig, ax = plt.subplots(figsize=(8, 4))
sns.kdeplot(data=df_tips, x="total_bill", fill=True, color="teal", ax=ax)
ax.set_title("Total Bill Kernel Density Estimation (KDE Plot)")
plt.show()
````

````python
# Histogram and KDE
plt.figure(figsize=(12, 5))

# Subplots
plt.subplot(1, 2, 1)
sns.histplot(data=tips, x="total_bill", bins=20, kde=True, color="skyblue")
plt.title("Histogram with KDE")

plt.subplot(1, 2, 2)
sns.kdeplot(
    data=tips, x="total_bill", hue="time", fill=True, palette="pastel", alpha=0.6
)
plt.title("KDE Plot (by Time)")

plt.tight_layout()
plt.show()
````

# 4.5 Box Plot and Violin Plot

Box Plot: Summarizes data via median, quartiles, and outliers.
`sns.boxplot(x=category, y=value)`

Violin Plot: Provides same summary info as box plot but also shows distribution shape (density estimation) on sides. Offers more information than Box Plot.

Command: `sns.violinplot(x=category, y=value)`

````python
# Data: total_bill distribution by time (Lunch vs Dinner)
fig, ax = plt.subplots(figsize=(8, 5))

sns.boxplot(
    data=df_tips,
    x="time",  # Categorical variable (X axis)
    y="total_bill",  # Numerical variable (Y axis)
    palette="Set2",  # Color palette
    ax=ax,
)

ax.set_title("Lunch vs Dinner Bill Comparison (Box Plot)")
ax.set_xlabel("Meal Time")
ax.set_ylabel("Total Bill ($)")

plt.show()
````

````python
# Violin

fig, ax = plt.subplots(figsize=(8, 5))

sns.violinplot(
    data=df_tips,
    x="day",  # Days of week
    y="tip",  # Tip amount
    palette="viridis",
    ax=ax,
)

ax.set_title("Tip Distribution by Day of Week (Violin Plot)")
ax.set_xlabel("Day of Week")
ax.set_ylabel("Tip Amount ($)")

plt.show()
````

````python
# Box plot vs violin plot comparison
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(data=tips, x="day", y="total_bill", hue="smoker", palette="Set2")
plt.title("Box Plot")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")

plt.subplot(1, 2, 2)
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker",
    split=True,
    palette="Set2",
    inner="quartile",
)
plt.title("Violin Plot")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
````

# 7. Style and Appearance Customization
# 7.1 Themes and Color Palettes

Unlike Matplotlib's scatter, Seaborn's scatterplot allows visualizing additional dimensions using extra categorical variables (e.g., color, size, or shape).

````python
# Reload dataset
df_tips = sns.load_dataset("tips")

fig, ax = plt.subplots(figsize=(10, 6))

# Plot relationship between total_bill (X) and tip (Y)
# hue: Color points by 'time' category
# style: Change marker shape by 'smoker' category
sns.scatterplot(
    data=df_tips,
    x="total_bill",
    y="tip",
    hue="time",  # Show 3rd variable via color
    style="smoker",  # Show 4th variable via shape
    s=80,  # Marker size
    ax=ax,
)

ax.set_title("Total Bill vs Tip Relationship (by Time and Smoker Status)")
ax.legend(title="Meal Time", loc="upper left")

plt.show()
````

````python
# Trying different themes
themes = ["darkgrid", "whitegrid", "dark", "white", "ticks"]
palettes = ["deep", "muted", "pastel", "bright", "dark", "colorblind"]

# Color palette examples
plt.figure(figsize=(15, 10))

for i, (theme, palette) in enumerate(zip(themes[:3], palettes[:3])):
    sns.set_theme(style=theme)

    plt.subplot(2, 3, i + 1)
    sns.scatterplot(
        data=iris, x="sepal_length", y="sepal_width", hue="species", palette=palette
    )
    plt.title(f"Theme: {theme}, Palette: {palette}")

sns.set_theme(style="whitegrid")  # Return to default theme
plt.tight_layout()
plt.show()
````

# 7.2 Detailed Customization

````python
# Detailed customization example
with sns.axes_style("whitegrid"):
    plt.figure(figsize=(12, 8))

    # Custom color palette
    custom_palette = sns.color_palette("husl", 3)

    # Violin plot with swarm plot
    ax = sns.violinplot(
        data=iris,
        x="species",
        y="petal_length",
        palette=custom_palette,
        inner=None,
        alpha=0.7,
    )
    sns.swarmplot(
        data=iris, x="species", y="petal_length", color="black", alpha=0.8, size=3
    )

    # Plot customization
    ax.set_title(
        "Petal Length Distribution by Iris Species",
        fontsize=16,
        fontweight="bold",
        pad=20,
    )
    ax.set_xlabel("Iris Species", fontsize=12, fontweight="bold")
    ax.set_ylabel("Petal Length (cm)", fontsize=12, fontweight="bold")

    # Remove spines
    sns.despine(left=True, bottom=True)

plt.tight_layout()
plt.show()
````

# 8. Regression Plots
# 8.1 Reg Plot
These plots show linear relationship between two variables while automatically adding best-fit regression line and confidence interval.

````python
fig, ax = plt.subplots(figsize=(8, 5))

# Show linear relationship between total_bill and tip
# ci=95: Show 95% confidence interval (default)
sns.regplot(
    data=df_tips,
    x="total_bill",
    y="tip",
    scatter_kws={"color": "teal", "alpha": 0.7},  # Point settings
    line_kws={"color": "red"},  # Line settings
    ax=ax,
)

ax.set_title("Tip Prediction with Regression Line")
plt.show()
````

````python
# Regression plots
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    scatter_kws={"alpha": 0.5},
    line_kws={"color": "red"},
)
plt.title("Simple Linear Regression")

plt.subplot(1, 3, 2)
sns.regplot(data=tips, x="total_bill", y="tip", scatter_kws={"alpha": 0.5})
plt.title("LOWESS Regression")

plt.subplot(1, 3, 3)
sns.residplot(data=tips, x="total_bill", y="tip", scatter_kws={"alpha": 0.5})
plt.title("Residual Plot")

plt.tight_layout()
plt.show()
````

# 9.1 Facet Grid
FacetGrid or its high-level counterparts (relplot, catplot) split a large dataset by values of one or two categorical variables and draw same plot for each subset. This allows quick comparison of differences between subgroups.

````python
# Example: Tip Distribution by Size Across Days
# sns.catplot places a categorical plot type (e.g., boxplot) into FacetGrid.
sns.catplot(
    data=df_tips,
    x="time",  # X axis: Meal Time
    y="tip",  # Y axis: Tip Amount
    kind="box",  # Plot type (boxplot, violin, bar etc.)
    col="day",  # Split columns by 'day' category
    col_wrap=2,  # Limit max columns to 2
    height=4,  # Height of each subplot
    aspect=1.2,  # Aspect ratio of each subplot
    palette="muted",
)

plt.suptitle("Meal Time and Tip Distribution (Separated by Day)", y=1.02)
plt.show()
````
