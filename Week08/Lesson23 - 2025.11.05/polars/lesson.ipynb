# **Polars DataFrame**

DataFrame is the most important data structure of the Polars library. It is a 2-dimensional, typed data structure like an Excel table.

DataFrames are structures formed by the combination of equal-length Series. They are often likened to SQL tables with columns and rows of different types. DataFrames allow us to process data more easily. They consist of named Columns and rows; rows are identified by position only (there is **no index**). Data analysis can be performed on these values with expressions.

Two important differences from Pandas:

* **No index**: if you need a row label, keep it in a normal column.
* **Immutable style**: methods such as `with_columns`, `filter`, `sort` return a new DataFrame, so assign the result (`df = df.with_columns(...)`) if you want to keep it.

# Accessing General DataFrame Properties

````python
print("\n" + "=" * 50)
print("BASIC DATAFRAME PROPERTIES")
print("=" * 50)

data = {
    "Name": ["Jack", "Jim", "John", "Jill", "Jenny"],
    "Age": [25, 30, 35, 28, 32],
    "City": ["New York", "Boston", "Los Angles", "Miami", "Chicago"],
    "Salary": [50000, 60000, 70000, 55000, 65000],
}

df = pl.DataFrame(data)

print("DataFrame info:")
df.glimpse()  # Polars' counterpart of df.info() (prints directly)

print(f"\nSize: {df.shape}")
print(f"Row count: {len(df)}")
print(f"Column count: {len(df.columns)}")
print(f"Column names: {df.columns}")
print("Index info: Polars has no index (rows are addressed by position)")
print(f"Data types:\n{df.schema}")

print("\nFirst 3 rows:")
print(df.head(3))

print("\nLast 2 rows:")
print(df.tail(2))
````

# Column Operations

````python
print("\n" + "=" * 50)
print("COLUMN OPERATIONS")
print("=" * 50)

# Selecting column
print("Name column:")
print(df["Name"])
print(type(df["Name"]))  # polars Series

# Selecting multiple columns
print("\nName and Age columns:")
print(df.select("Name", "Age"))

# Adding new column (with_columns returns a NEW DataFrame, so we reassign)
df = df.with_columns(
    pl.Series("Experience", [3, 5, 8, 4, 6]),
    (pl.col("Salary") * 12).alias("Annual Salary"),
    pl.when(pl.col("Age") < 30)
    .then(pl.lit("Young"))
    .otherwise(pl.lit("Adult"))
    .alias("Age Group"),  # np.where equivalent
)

print("\nNew columns added:")
print(df)

# Deleting column
df_drop = df.drop("Age Group")  # no axis argument needed
print("\nAge Group column dropped:")
print(df_drop.columns)
````

# Row Operations

````python
print("\n" + "=" * 50)
print("ROW OPERATIONS")
print("=" * 50)

# Selecting row by position (there is no .loc/.iloc)
print("First row (as a tuple / as a dict):")
print(df.row(0))
print(df.row(0, named=True))

print("\nFirst 3 rows:")
print(df[:3])  # same as df.head(3) / df.slice(0, 3)

print("\nRow at specific position (3rd row):")
print(df[2])  # a one-row DataFrame

# Selecting rows by condition
print("\nThose older than 30:")
print(df.filter(pl.col("Age") > 30))

print("\nThose living in Istanbul:")
print(df.filter(pl.col("City") == "Istanbul"))

print("\nThose living in Boston:")
print(df.filter(pl.col("City") == "Boston"))

# Multiple condition (several predicates are AND-ed)
print("\nThose older than 30 AND salary more than 60000:")
print(df.filter(pl.col("Age") > 30, pl.col("Salary") > 60000))
````

# Row Labels Instead of an Index

Polars has no index. The common index tasks map to:

* an auto-incrementing id → `with_row_index`
* a label/key → a normal column (use `filter`, `join` or `partition_by` to look it up)

````python
print("\n" + "=" * 50)
print("INDEX OPERATIONS (no index in Polars)")
print("=" * 50)

# "Setting custom index": add the labels as a column
df_custom = df.with_columns(pl.Series("label", ["A", "B", "C", "D", "E"]))
print("DataFrame with a label column:")
print(df_custom)

# "Resetting index": a 0..n-1 counter column (UInt32)
df_reset = df_custom.drop("label").with_row_index()
print("\nRow counter added:")
print(df_reset)

# "Making index a column": rename the counter to the name you want
df_custom = df_custom.drop("label").with_row_index("ID")
print("\nCounter as ID column:")
print(df_custom)

# "Making a column the index": no need, but for fast key lookups you can split by key
by_name = df.partition_by("Name", as_dict=True, include_key=True)
print("\nLookup by Name (dict of DataFrames):")
print(by_name[("John",)])
````

# Data Filtering and Querying

````python
print("\n" + "=" * 50)
print("DATA FILTERING AND QUERYING")
print("=" * 50)

# .filter() method
print("Querying with .filter():")
print(df.filter((pl.col('Age') > 30) & (pl.col('Salary') > 60000)))

# Filtering with .isin()
cities = ["Boston", "New York"]
print(f"\nThose living in {cities} cities:")
print(df.filter(pl.col("City").is_in(cities)))

# Filtering with string operations
print("\nNames starting with 'A':")
print(df.filter(pl.col('Name').str.starts_with("A")))

print("\nNames starting with 'J':")
print(df.filter(pl.col('Name').str.starts_with("J")))

# Range querying with Between
print("\nThose aged between 25-35:")
print(df.filter(pl.col('Age').is_between(25,35)))
````

# Data Grouping and Aggregation

````python
print("\n" + "=" * 50)
print("DATA GROUPING AND AGGREGATION")
print("=" * 50)

# Grouping operations
# (group_by does not guarantee row order, so we sort for a stable output)
print("Average salary by city:")
print(df.group_by("City").agg(pl.col("Salary").mean()).sort("City"))

print("\nDetailed statistics by city:")
print(
    df.group_by("City")
    .agg(
        pl.col("Salary").mean().alias("Salary_mean"),
        pl.col("Salary").min().alias("Salary_min"),
        pl.col("Salary").max().alias("Salary_max"),
        pl.col("Salary").std().alias("Salary_std"),
        pl.col("Age").mean().alias("Age_mean"),
        pl.col("Age").count().alias("Age_count"),
    )
    .sort("City")
)

# Multiple grouping
df = df.with_columns(pl.Series("Gender", ["M", "M", "M", "F", "F"]))
print("\nGrouping by City and Gender:")
print(df.group_by("City", "Gender").agg(pl.col("Salary").mean()).sort("City", "Gender"))
````

# Missing Data Operations

In Polars missing data is `null` for every dtype. `NaN` is a regular floating point value and is **not** treated as missing by `is_null()` / `drop_nulls()`.

````python
print("\n" + "=" * 50)
print("MISSING DATA OPERATIONS")
print("=" * 50)

# Creating DataFrame with missing data (set the value at a given row position to null;
# pl.int_range(pl.len()) is the row position)
df_missing = df.with_columns(
    pl.when(pl.int_range(pl.len()) == 0)
    .then(None)
    .otherwise(pl.col("Age"))
    .alias("Age"),
    pl.when(pl.int_range(pl.len()) == 2)
    .then(None)
    .otherwise(pl.col("Salary"))
    .alias("Salary"),
    pl.when(pl.int_range(pl.len()) == 4)
    .then(None)
    .otherwise(pl.col("City"))
    .alias("City"),
)

print("DataFrame with missing data:")
print(df_missing)

print("\nMissing data check:")
print(df_missing.select(pl.all().is_null()))

print("\nMissing data counts:")
print(df_missing.null_count())

# Cleaning missing data
df_clean = df_missing.drop_nulls()  # Drop rows with missing data
print("\nMissing data dropped:")
print(df_clean)

# Filling missing data
df_filled = df_missing.with_columns(
    pl.col("Age").fill_null(pl.col("Age").mean()),
    pl.col("Salary").fill_null(pl.col("Salary").median()),
    pl.col("City").fill_null("Unknown"),
)
print("\nMissing data filled:")
print(df_filled)

# null vs NaN
s = pl.Series("x", [1.0, None, float("nan")])
print("\nnull vs NaN:")
print("is_null:", s.is_null().to_list())  # [False, True, False]
print("is_nan: ", s.is_nan().to_list())  # [False, None, True]
print("fill_nan(None) converts NaN to null:", s.fill_nan(None).is_null().to_list())
````

# DataFrame Merging

````python
print("\n" + "=" * 50)
print("DATAFRAME MERGING OPERATIONS")
print("=" * 50)

# Let's create two DataFrames
df1 = pl.DataFrame(
    {
        "ID": [1, 2, 3, 4],
        "Name": ["John", "Jill", "Jack", "Jenny"],
        "Department": ["IT", "HR", "IT", "Finance"],
    }
)

df2 = pl.DataFrame(
    {
        "ID": [1, 2, 3, 5],
        "Salary": [50000, 60000, 70000, 80000],
        "Experience": [3, 5, 8, 10],
    }
)

print("DF1:")
print(df1)
print("\nDF2:")
print(df2)

# Inner join
inner_merge = df1.join(df2, on="ID", how="inner")
print("\nInner Join:")
print(inner_merge)

# Left join
left_merge = df1.join(df2, on="ID", how="left")
print("\nLeft Join:")
print(left_merge)

# Full (outer) join: coalesce=True merges the two key columns into one
full_merge = df1.join(df2, on="ID", how="full", coalesce=True).sort("ID")
print("\nFull Join:")
print(full_merge)

# Concatenation
df3 = pl.DataFrame(
    {"ID": [5, 6], "Name": ["Jimmy", "Jony"], "Department": ["IT", "HR"]}
)

merged = pl.concat([df1, df3])  # no ignore_index needed: there is no index
print("\nMerging with Concat:")
print(merged)
````

# A General Example

````python
print("\n" + "=" * 50)
print("REAL LIFE EXAMPLE: Employee Analysis")
print("=" * 50)

# Detailed analysis (expressions inside the same with_columns can't see each other's new columns,
# so Net Salary is created in a second, chained with_columns)
df = df.with_columns(
    (pl.col("Salary") * 0.15).alias("Tax"),
    pl.Series("Performance", ["A", "B", "A", "C", "B"]),
).with_columns((pl.col("Salary") - pl.col("Tax")).alias("Net Salary"))

print("Detailed Employee DataFrame:")
print(df)

# Detailed analysis
analysis = (
    df.group_by("Performance")
    .agg(
        pl.col("Salary").count().alias("Salary_count"),
        pl.col("Salary").mean().alias("Salary_mean"),
        pl.col("Salary").std().alias("Salary_std"),
        pl.col("Age").min().alias("Age_min"),
        pl.col("Age").max().alias("Age_max"),
        pl.col("Age").mean().alias("Age_mean"),
        pl.col("Net Salary").sum().alias("Net Salary_sum"),
    )
    .sort("Performance")
    .with_columns(cs.float().round(2))
)

print("\nPerformance Analysis:")
print(analysis)

# Correlation analysis (cs.numeric() selects all numeric columns)
numeric_df = df.select(cs.numeric())
print("\nCorrelation Matrix:")
print(numeric_df.corr())
````

# **Important Features of DataFrame:**
2-Dimensional: Consists of rows and columns

Flexible Data Types: Can hold data of different types (strictly typed per column)

Named Access: Named columns, rows addressed by position (no index)

Missing Data Support: Can manage null values

````python
# scenario 2 : data quality control

print("\n" + "=" * 50)
print("SCENARIO 2: DATA QUALITY CONTROL")
print("=" * 50)


def data_quality_control(dataframe):
    """Check data quality"""
    print("✅ DATA QUALITY REPORT")
    print("=" * 35)

    # Size info
    print(f"Row count: {len(dataframe)}")
    print(f"Column count: {len(dataframe.columns)}")

    # Missing data analysis
    missing_data = dataframe.null_count().row(0, named=True)
    print(f"\nMISSING DATA ANALYSIS:")
    for column, missing in missing_data.items():
        if missing > 0:
            print(f"  ⚠️  {column}: {missing} missing data")
        else:
            print(f"  ✅ {column}: No missing data")

    # Data types
    print(f"\nDATA TYPES:")
    for column, tip in dataframe.schema.items():
        print(f"  {column}: {tip}")

    # Unique values
    print(f"\nUNIQUE VALUE COUNTS:")
    for column in dataframe.columns:
        unique = dataframe[column].n_unique()
        print(f"  {column}: {unique} unique values")


data_quality_control(df)
````

# Polars File Operations

**csv and excel file operations**

Polars has two ways of reading files:

* `pl.read_*` (eager): loads the file into a DataFrame immediately.
* `pl.scan_*` (lazy): returns a LazyFrame; the query is optimized and only the needed rows/columns are read. This is the preferred way for large files.

There is no index in Polars, so there are no `index=` / `index_col=` parameters when writing or reading.

# **Writing CSV (write_csv)**

````python
# Basic writing (there is no index, so nothing extra is written)
df.write_csv("file.csv")

# Writing with parameters
df.write_csv(
    "file_with_parameters.csv",
    separator=";",  # Change separator
    include_header=True,  # Write header row
)
# Note: write_csv has no `encoding` and no `index` parameter.
# Polars always writes UTF-8 and never writes an index.
````

# **Reading CSV (read_csv)**

````python
# Basic reading
df2 = pl.read_csv("file.csv")
print(df2.head(3))

# Reading with parameters
df3 = pl.read_csv(
    "file.csv",
    encoding="utf8",  # Encoding type ("utf8" or "utf8-lossy")
    separator=",",  # Separator (delimiter)
    has_header=True,  # First row is the header (pandas: header=0)
    columns=["Name", "Age"],  # Read only specific columns (pandas: usecols)
    n_rows=1000,  # Read first 1000 rows (pandas: nrows)
    skip_rows_after_header=5,  # Skip 5 rows after the header (skip_rows skips BEFORE the header)
    null_values=["", "NULL", "NaN"],  # Values to consider as null (pandas: na_values)
)
# There is no index_col: Polars has no index, the column simply stays a column.
print(df3)
````

**Writing Excel (write_excel)**

````python
# Basic writing (the sheet is called "Sheet1" by default)
df.write_excel("file.xlsx")

# Writing with parameters
df.write_excel(
    "file_with_parameters.xlsx",
    worksheet="Data",  # Sheet name
    include_header=True,  # Write header row
    position="B2",  # Top-left cell (pandas: startrow=1, startcol=1)
)

# Writing multiple DataFrames to different sheets:
# create one xlsxwriter Workbook and pass it to every write_excel call
from xlsxwriter import Workbook

with Workbook("multi_sheet.xlsx") as workbook:
    df1.write_excel(workbook, worksheet="Sheet1")
    df2.write_excel(workbook, worksheet="Sheet2")
    df3.write_excel(workbook, worksheet="Sheet3")
````

**Reading Excel (read_excel)**

````python
# Basic reading (first sheet)
df = pl.read_excel("file.xlsx")

# Reading specific sheet
df = pl.read_excel("file.xlsx", sheet_name="Sheet1")

# Reading with parameters (stored in a separate variable, df keeps all columns)
df_subset = pl.read_excel(
    "file.xlsx",
    sheet_id=1,  # Sheet number, 1-based (or use sheet_name="...")
    engine="calamine",  # fastexcel (default engine)
    read_options={  # options of the underlying engine
        "header_row": 0,  # Header row
        "use_columns": "A:C,E",  # Read specific columns
        "n_rows": 1000,  # Row count limit
    },
)
# Custom null markers (pandas: na_values): replace them after reading
df_subset = df_subset.with_columns(pl.col(pl.String).replace(["", "NULL"], None))
# Skipping rows after the header: df_subset = df_subset.slice(5)
print(df_subset.columns)

# Reading all sheets: sheet_id=0 returns a dict {sheet name: DataFrame}
all_sheets = pl.read_excel("multi_sheet.xlsx", sheet_id=0)
print(list(all_sheets.keys()))

# Getting the sheet names first (pandas: pd.ExcelFile(...).sheet_names)
import fastexcel

excel_file = fastexcel.read_excel("multi_sheet.xlsx")
sheet_names = excel_file.sheet_names

# Reading each sheet as separate DataFrame
dfs = {}
for sheet in sheet_names:
    dfs[sheet] = pl.read_excel("multi_sheet.xlsx", sheet_name=sheet)
print({name: frame.shape for name, frame in dfs.items()})
````

# *Data Cleaning*

Polars objects are immutable: these calls return a **new** DataFrame, so assign the result if you want to keep it.
(The `df` below has the columns `Name`, `Age`, `City`, `Salary`, `Department`, `Experience`.)

````python
# Handling missing data
df.drop_nulls()  # Drop rows containing null
df.fill_null(0)  # Fill null values with 0 (applied to the columns where 0 fits)
df.fill_null(
    strategy="forward"
)  # Fill with previous value (pandas: fillna(method="ffill"))

# Column operations
df.rename({"Experience": "Years of Experience"})  # Rename column
df.drop("Department")  # Drop column
df = df.with_columns(
    (pl.col("Salary") / pl.col("Experience")).alias("Salary per Year")
)  # Create new column

# Data type conversion
df = df.with_columns(pl.col("Experience").cast(pl.Int32))

# Date conversion (pd.to_datetime)
dates_df = pl.DataFrame({"date": ["2023-01-15", "2023-02-20", "2023-03-25"]})
dates_df = dates_df.with_columns(pl.col("date").str.to_date())
print(dates_df)
````

# File Reading Example

# **Let's create a sample CSV file that we can use to test polars operations**

````python
# Random data creation functions
def random_date(start=datetime(2023, 1, 1), end=datetime(2024, 1, 1)):
    """Create random date"""
    diff = end - start
    random_day = random.randint(0, diff.days)
    return start + timedelta(days=random_day)


def random_product():
    """Create random product name"""
    products = [
        "Laptop",
        "Phone",
        "Tablet",
        "Headphone",
        "Mouse",
        "Keyboard",
        "Monitor",
        "Printer",
        "Camera",
        "Speaker",
        "Hard Disk",
        "SSD",
    ]
    return random.choice(products)


def random_city():
    """Create random city name"""
    cities = [
        "Istanbul",
        "Ankara",
        "Izmir",
        "Bursa",
        "Antalya",
        "Adana",
        "Konya",
        "Gaziantep",
        "Kayseri",
        "Mersin",
        "Diyarbakir",
        "Eskisehir",
    ]
    return random.choice(cities)


def random_category(product):
    """Determine category based on product"""
    category_map = {
        "Laptop": "Electronics",
        "Phone": "Electronics",
        "Tablet": "Electronics",
        "Headphone": "Accessory",
        "Mouse": "Accessory",
        "Keyboard": "Accessory",
        "Monitor": "Hardware",
        "Printer": "Hardware",
        "Camera": "Electronics",
        "Speaker": "Accessory",
        "Hard Disk": "Hardware",
        "SSD": "Hardware",
    }
    return category_map.get(product, "Other")


# Create dataset
np.random.seed(42)  # For reproducibility
random.seed(42)

sales = []

for i in range(1000):
    product = random_product()
    category = random_category(product)
    date = random_date()
    city = random_city()
    quantity = random.randint(1, 5)
    unit_price = round(random.uniform(50, 2000), 2)
    total_amount = round(quantity * unit_price, 2)

    # Add missing data in some rows (for realism)
    # In Polars missing data is None (-> null), not np.nan
    customer_age = random.randint(18, 70) if random.random() > 0.1 else None
    customer_gender = (
        random.choice(["Male", "Female"]) if random.random() > 0.08 else None
    )

    sales.append(
        {
            "sale_id": i + 1,
            "customer_id": f"M{random.randint(1000, 9999)}",
            "product_name": product,
            "category": category,
            "quantity": quantity,
            "unit_price": unit_price,
            "total_amount": total_amount,
            "date": date.strftime("%Y-%m-%d"),
            "city": city,
            "customer_age": customer_age,
            "customer_gender": customer_gender,
            "payment_type": random.choice(["Credit Card", "Cash", "Transfer"]),
            "discount_rate": round(random.uniform(0, 0.3), 2),  # 0-30% discount
        }
    )

# Create DataFrame
# infer_schema_length=None: look at ALL rows to infer dtypes (default is only the first 100 rows)
df = pl.DataFrame(sales, infer_schema_length=None)

# Save to CSV file
df.write_csv("sales_data.csv")

print("CSV file created successfully!")
print(f"Created dataset size: {df.shape}")
print("\nFirst 5 rows:")
print(df.head())
print("\nData types:")
print(df.schema)
print("\nBasic statistics:")
print(df.describe())
````
